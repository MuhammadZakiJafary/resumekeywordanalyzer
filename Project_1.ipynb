```python
import re
```

```python
skills = [
    "python",
    "sql",
    "machine learning",
    "deep learning",
    "data analysis",
    "data science",
    "pandas",
    "numpy",
    "scikit-learn",
    "tensorflow",
    "keras",
    "pytorch",
    "computer vision",
    "natural language processing",
    "nlp",
    "opencv",
    "flask",
    "streamlit",
    "docker",
    "aws",
    "git",
    "github",
    "java",
    "c++",
    "javascript",
    "html",
    "css"
]
```

```python
resume = """
I am a Software Engineering graduate with knowledge of Python,
SQL, Machine Learning, Pandas, NumPy, Scikit-learn and
Computer Vision. I have worked on Machine Learning and
Deep Learning projects.
"""

job_description = """
We are looking for a Machine Learning Intern with knowledge of
Python, SQL, Machine Learning, Pandas, Scikit-learn,
TensorFlow, Docker and AWS.
"""
```

```python
def preprocess_text(text):

    # Convert text to lowercase
    text = text.lower()

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text)

    return text
```

```python
def extract_skills(text, skills_list):

    found_skills = []

    for skill in skills_list:

        if skill in text:
            found_skills.append(skill)

    return found_skills
```

```python
clean_resume = preprocess_text(resume)
clean_job_description = preprocess_text(job_description)

resume_skills = extract_skills(clean_resume, skills)

job_skills = extract_skills(clean_job_description, skills)

matched_skills = list(set(resume_skills) & set(job_skills))

missing_skills = list(set(job_skills) - set(resume_skills))
```

```python
if len(job_skills) > 0:
    coverage = (len(matched_skills) / len(job_skills)) * 100
else:
    coverage = 0
```

```python
print("=" * 50)
print("RESUME KEYWORD ANALYSIS REPORT")
print("=" * 50)

print("\nTotal Required Skills:", len(job_skills))
print("Matched Skills:", len(matched_skills))
print("Missing Skills:", len(missing_skills))

print(f"\nKeyword Coverage: {coverage:.2f}%")

print("\nMATCHED SKILLS")

if matched_skills:
    for skill in matched_skills:
        print("✓", skill.title())
else:
    print("No matched skills found.")


print("\nMISSING SKILLS")

if missing_skills:
    for skill in missing_skills:
        print("✗", skill.title())
else:
    print("No missing skills found.")
```
